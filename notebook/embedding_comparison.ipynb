# Embedding 설정 비교 기록

이 노트북은 현재 환경에서 선택된 임베딩 제공자·모델·벡터 차원을 확인하는 설정 점검용입니다.
이 셀은 실제 임베딩 API를 호출하지 않으며, 민감한 API 키를 출력하지 않습니다. 실제 provider 비교는 비용과 호출 권한을 확인한 뒤 별도로 수행합니다.

```
# 환경 변수와 기본값을 통합한 프로젝트 설정을 불러옵니다.
from app.core.config import get_settings

# 설정 객체에서 임베딩 관련 공개 설정만 선택합니다.
settings = get_settings()
{
    "provider": settings.embedding_provider,
    "model": settings.embedding_model,
    # pgvector 컬럼 차원과 일치해야 하는 값입니다.
    "dimension": settings.embedding_dimension,
}
```
