# 승인 로컬 데이터셋 EDA

이 노트북은 운영 적재 로직을 다시 구현하지 않고, 이미 정의된 manifest loader를 사용하여 승인된 로컬 문서의 기본 상태를 확인합니다.
문서 수·본문 길이·검수 상태를 확인한 뒤, 실제 ingestion을 실행하기 전에 데이터 준비 상태를 점검하는 것이 목적입니다.

```
# manifest 파일의 위치를 지정합니다.
from pathlib import Path

from app.ingestion.document_loader import load_manifest

# 운영 코드와 동일한 loader를 사용하여 데이터 구조를 일관되게 검증합니다.
manifest = Path("../data/manifest.json")
if not manifest.exists():
    raise FileNotFoundError(
        "manifest가 없습니다. 승인 문서와 data/manifest.json을 먼저 준비하세요."
    )
documents = load_manifest(manifest)

# 문서 ID, 본문 길이, 검수 상태를 요약합니다.
[(d.document_id, len(d.content), d.review_status) for d in documents]
```
