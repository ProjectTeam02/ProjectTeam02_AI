# Single-turn 대 Multi-turn 검색 비교

동일한 질문과 정답 문서 집합을 기준으로, 대화 맥락을 사용하지 않은 검색과 이전 대화 맥락을 반영한 검색을 비교합니다.
실제 서비스 성능을 의미하려면 여러 질문과 사람이 작성한 정답 문서 집합이 필요하며, 아래 코드는 평가 함수의 동작 예시입니다.

```
# 두 검색 실행 결과를 같은 평가 함수로 비교합니다.
from app.evaluation.rag_metrics import compare_runs

# Single-turn은 관련 문서 DOC-RATE를 찾지 못한 예시입니다.
single_turn = [(["DOC-OTHER"], {"DOC-RATE"})]
# Multi-turn은 대화 맥락을 이용해 관련 문서를 찾은 예시입니다.
multi_turn = [(["DOC-RATE"], {"DOC-RATE"})]

# 상위 1개 검색 결과의 Recall과 MRR을 비교합니다.
compare_runs(single_turn, multi_turn, k=1)
```
